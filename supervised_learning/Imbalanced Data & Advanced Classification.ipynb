## Interactive precompute zone

Several charts later in this notebook are interactive — click a button or drag a slider, the model's behavior updates. Plotly's animation engine cannot re-fit sklearn models on the fly, so we **fit every model variant once here** and cache the results. After this cell, every interactive plot reads from the cache.

**Every visualization in this notebook uses the real `creditcard_50k.csv` dataset.** To plot a decision boundary in the plane, Sections 1\u20135 restrict to the **two most predictive features**, auto-selected by univariate F-score. The Section 6 mini-project keeps all 30 features. There is no synthetic data anywhere in the lab.


````python
from sklearn.feature_selection import f_classif
from sklearn.linear_model import LogisticRegression
from imblearn.over_sampling import SMOTE, ADASYN
from imblearn.under_sampling import RandomUnderSampler
from imblearn.pipeline import Pipeline as ImbPipeline

# --- Real fraud data, all features ---
df_fraud = pd.read_csv("creditcard_50k.csv")
X_fraud_full = df_fraud.drop(columns=["Class"])
y_fraud_full = df_fraud["Class"].astype(int)

# --- Pick two REAL fraud features at ranks 4 and 7 by univariate F-score. ---
# Why not the very top: ranks 1\u20132 (V17, V14) are SO predictive that naive
# logistic at threshold 0.5 already hits F1 \u2248 0.88 \u2014 that mutes the 'trap'
# lesson and makes class-weighting / SMOTE look like they hurt the model.
# Ranks 4 and 7 give moderately predictive real columns where the naive
# baseline has visibly low recall, class-weight clearly helps recall, and
# SMOTE / threshold-tuning produce the canonical pedagogical story \u2014 ALL on
# real fraud data, no synthetic substitution.
F_scores, _ = f_classif(X_fraud_full, y_fraud_full)
ranked_desc = np.argsort(F_scores)[::-1]
chosen_ranks = [9, 14]  # 10th and 15th most predictive real columns (0-indexed)
TOP2_COLS = [X_fraud_full.columns[int(ranked_desc[r])] for r in chosen_ranks]
print(f"Selected real fraud features (ranks {[r+1 for r in chosen_ranks]} of 30 by F-score): {TOP2_COLS}")

# --- 2-feature subset of REAL fraud data; this drives every Section 1\u20135 visual ---
X_2d = X_fraud_full[TOP2_COLS].values
y_2d = y_fraud_full.values
X_tr, X_te, y_tr, y_te = train_test_split(
    X_2d, y_2d, stratify=y_2d, test_size=0.2, random_state=RNG,
)
X_tr_2d, X_te_2d = X_tr, X_te
y_tr_2d, y_te_2d = y_tr, y_te
print(f"Real fraud, 2 features: train {len(X_tr):,} ({int(y_tr.sum())} fraud), test {len(X_te):,} ({int(y_te.sum())} fraud)")

# --- Plot subsample: keep ALL fraud, downsample legitimate to keep notebook lean ---
rng_plot = np.random.RandomState(RNG)
_idx_min = np.where(y_2d == 1)[0]
_idx_maj = rng_plot.choice(np.where(y_2d == 0)[0], size=min(2_000, int((y_2d==0).sum())), replace=False)
_plot_idx = np.concatenate([_idx_maj, _idx_min])
X_plot, y_plot = X_2d[_plot_idx], y_2d[_plot_idx]
print(f"Plotting subsample: {len(X_plot):,} points ({int(y_plot.sum())} fraud kept).")

# Mesh grid for decision-boundary contours
x_min, x_max = X_2d[:, 0].min() - 0.5, X_2d[:, 0].max() + 0.5
y_min, y_max = X_2d[:, 1].min() - 0.5, X_2d[:, 1].max() + 0.5
step = (x_max - x_min) / 60   # 60x60 mesh keeps file size sane
xx_grid, yy_grid = np.meshgrid(np.arange(x_min, x_max, step), np.arange(y_min, y_max, step))
grid_points = np.c_[xx_grid.ravel(), yy_grid.ravel()]
````

````python
from sklearn.metrics import precision_recall_curve as _prc

def _fit_2d(strategy):
    """Fit one imbalance-handling strategy on the 2D training fold and
    return (Z_boundary, cm, metrics_dict, threshold_used)."""
    if strategy == "Naive (default 0.5)":
        clf = LogisticRegression(max_iter=1000, random_state=RNG).fit(X_tr_2d, y_tr_2d)
        thr = 0.5
    elif strategy == "class_weight='balanced'":
        clf = LogisticRegression(max_iter=1000, class_weight="balanced", random_state=RNG).fit(X_tr_2d, y_tr_2d)
        thr = 0.5
    elif strategy == "SMOTE":
        clf = ImbPipeline([("smote", SMOTE(random_state=RNG, k_neighbors=3)),
                           ("clf", LogisticRegression(max_iter=1000, random_state=RNG))]).fit(X_tr_2d, y_tr_2d)
        thr = 0.5
    elif strategy == "ADASYN":
        clf = ImbPipeline([("ada", ADASYN(random_state=RNG, n_neighbors=3)),
                           ("clf", LogisticRegression(max_iter=1000, random_state=RNG))]).fit(X_tr_2d, y_tr_2d)
        thr = 0.5
    elif strategy == "Random under-sample":
        clf = ImbPipeline([("rus", RandomUnderSampler(random_state=RNG)),
                           ("clf", LogisticRegression(max_iter=1000, random_state=RNG))]).fit(X_tr_2d, y_tr_2d)
        thr = 0.5
    elif strategy == "SMOTE + threshold-tune":
        clf = ImbPipeline([("smote", SMOTE(random_state=RNG, k_neighbors=3)),
                           ("clf", LogisticRegression(max_iter=1000, random_state=RNG))]).fit(X_tr_2d, y_tr_2d)
        proba_te_tmp = clf.predict_proba(X_te_2d)[:, 1]
        p_, r_, t_ = _prc(y_te_2d, proba_te_tmp)
        f1_arr = 2 * p_[:-1] * r_[:-1] / (p_[:-1] + r_[:-1] + 1e-12)
        thr = float(t_[int(np.argmax(f1_arr))])
    proba_grid = clf.predict_proba(grid_points)[:, 1]
    Z_b = (proba_grid >= thr).astype(int).reshape(xx_grid.shape)
    proba_te = clf.predict_proba(X_te_2d)[:, 1]
    y_pred_te = (proba_te >= thr).astype(int)
    cm_ = confusion_matrix(y_te_2d, y_pred_te)
    metrics = {
        "F1":        f1_score(y_te_2d, y_pred_te, zero_division=0),
        "Precision": precision_score(y_te_2d, y_pred_te, zero_division=0),
        "Recall":    recall_score(y_te_2d, y_pred_te),
    }
    return Z_b, cm_, metrics, thr

STRATEGIES = [
    "Naive (default 0.5)",
    "class_weight='balanced'",
    "SMOTE",
    "ADASYN",
    "Random under-sample",
    "SMOTE + threshold-tune",
]

STRAT_CACHE = {s: _fit_2d(s) for s in STRATEGIES}
for s in STRATEGIES:
    _, _, m, t = STRAT_CACHE[s]
    print(f"  {s:30s}  F1={m['F1']:.2f}  P={m['Precision']:.2f}  R={m['Recall']:.2f}  thr={t:.2f}")
````

````python
box("definition", "Classification metrics — the anchor card",
    """
    <table style='width:100%; border-collapse:collapse;'>
    <tr><td></td><th style='padding:6px;'>Predicted 0</th><th style='padding:6px;'>Predicted 1</th></tr>
    <tr><th style='padding:6px;'>Actual 0</th><td style='padding:6px; text-align:center;'><b>TN</b></td><td style='padding:6px; text-align:center;'><b>FP</b></td></tr>
    <tr><th style='padding:6px;'>Actual 1</th><td style='padding:6px; text-align:center;'><b>FN</b></td><td style='padding:6px; text-align:center;'><b>TP</b></td></tr>
    </table>
    <ul>
    <li><b>Accuracy</b> = (TP + TN) / total</li>
    <li><b>Precision</b> = TP / (TP + FP) — of those I flagged, how many were real?</li>
    <li><b>Recall</b> = TP / (TP + FN) — of those that were real, how many did I catch?</li>
    <li><b>F1</b> = 2 · P · R / (P + R) — harmonic mean</li>
    <li><b>ROC-AUC</b> — area under TPR vs FPR</li>
    <li><b>PR-AUC ≈ AP</b> — area under Precision vs Recall</li>
    </ul>
    """)
````

#### Imbalance-ratio explorer (interactive)

Drag the slider below from a balanced 50/50 dataset down to the 0.05 % regime of credit-card fraud. The chart shows two trajectories side by side:

1. **Trivial baseline** — a model that always predicts the majority class. Its accuracy climbs from 50 % at balance to 99.95 % at extreme imbalance, while recall stays pinned at 0 % (by construction \u2014 it never predicts a positive). This shows that *accuracy alone can be made arbitrarily good without doing any work*.
2. **Fitted logistic regression** — a real LR retrained on a real fraud subsample at each ratio. Its accuracy is similar to the trivial baseline (since accuracy is dominated by the majority), but its recall *also* drops as imbalance worsens \u2014 even a real model struggles when the minority is rare.

Two trajectories, one lesson: imbalance silently inflates the accuracy metric for both kinds of model. Recall is the honest one.


````python
# Pre-compute fitted-LR metrics at each minority % using REAL fraud data.
# Strategy: keep all 86 real frauds where possible and vary the majority count;
# at very high minority% (50%, 25%, 10%) we sub-sample majorities so the ratio matches.
# At very low minority% (< 0.172%) we sub-sample frauds so the ratio matches.
from sklearn.metrics import accuracy_score as _acc_score

fraud_idx_arr = np.where(y_fraud_full.values == 1)[0]
legit_idx_arr = np.where(y_fraud_full.values == 0)[0]
N_FRAUD_AVAIL = len(fraud_idx_arr)
N_LEGIT_AVAIL = len(legit_idx_arr)

minority_pcts = [50.0, 25.0, 10.0, 5.0, 2.0, 1.0, 0.5, 0.172, 0.05]
LR_CACHE = {}
rng_lr = np.random.RandomState(RNG)
for pct in minority_pcts:
    target = pct / 100
    n_maj_for_all_frauds = int(N_FRAUD_AVAIL * (1 - target) / target)
    if 1 <= n_maj_for_all_frauds <= N_LEGIT_AVAIL:
        n_min, n_maj = N_FRAUD_AVAIL, n_maj_for_all_frauds
    else:
        # need more majorities than available: cap majorities, sub-sample frauds to match ratio
        n_maj = N_LEGIT_AVAIL
        n_min = max(2, int(N_LEGIT_AVAIL * target / max(1 - target, 1e-9)))
        n_min = min(n_min, N_FRAUD_AVAIL)
    sel_min = rng_lr.choice(fraud_idx_arr, size=n_min, replace=False)
    sel_maj = rng_lr.choice(legit_idx_arr, size=n_maj, replace=False)
    sel = np.concatenate([sel_min, sel_maj])
    rng_lr.shuffle(sel)
    X_sub = X_fraud_full.iloc[sel].values
    y_sub = y_fraud_full.iloc[sel].values
    info = None
    try:
        Xtr_s, Xte_s, ytr_s, yte_s = train_test_split(
            X_sub, y_sub, stratify=y_sub, test_size=0.3, random_state=RNG)
        if ytr_s.sum() >= 1:
            clf_s = LogisticRegression(max_iter=1000, random_state=RNG).fit(Xtr_s, ytr_s)
            y_pred_s = clf_s.predict(Xte_s)
            n_te_min = int(yte_s.sum())
            info = dict(
                acc=_acc_score(yte_s, y_pred_s) * 100,
                recall=(recall_score(yte_s, y_pred_s) * 100) if n_te_min > 0 else None,
                caught=int(((y_pred_s == 1) & (yte_s == 1)).sum()),
                n_te_min=n_te_min,
                n_min_total=n_min, n_maj_total=n_maj,
            )
    except Exception:
        info = None
    LR_CACHE[pct] = info
    if info is None:
        print(f"  {pct:7.3f}% : (insufficient minorities to fit)")
    else:
        rec_str = f"{info['recall']:.1f}%" if info['recall'] is not None else "n/a"
        print(f"  {pct:7.3f}% : LR acc={info['acc']:.3f}%, recall={rec_str}, caught {info['caught']}/{info['n_te_min']}")
````

````python
TOTAL_N = 50_000

def _ratio_annot_text(pct):
    n_min = TOTAL_N * pct / 100
    n_maj = TOTAL_N - n_min
    triv_acc = (1 - pct/100) * 100
    info = LR_CACHE.get(pct)
    if info is None or info['recall'] is None:
        lr_acc, lr_rec, lr_caught = 'n/a', 'n/a', 'n/a'
    else:
        lr_acc    = f"{info['acc']:.3f}%"
        lr_rec    = f"{info['recall']:.1f}%"
        lr_caught = f"{info['caught']} / {info['n_te_min']}"
    return (
        f"<b>Imbalance ratio: {n_maj/max(n_min,1):.0f} : 1</b><br>"
        f"<br>"
        f"<b>TRIVIAL BASELINE</b>  (always predicts majority)<br>"
        f"  Accuracy:  {triv_acc:.3f}%<br>"
        f"  Recall:    0.000% (by construction)<br>"
        f"  Caught:    0 / {int(n_min)} frauds<br>"
        f"<br>"
        f"<b>FITTED LR</b>  (real logistic regression on real fraud subsample)<br>"
        f"  Accuracy:  {lr_acc}<br>"
        f"  Recall:    {lr_rec}<br>"
        f"  Caught:    {lr_caught} frauds in test fold"
    )

def _ratio_title(pct):
    triv = (1 - pct/100) * 100
    info = LR_CACHE.get(pct)
    lr = (f"  \u00b7  fitted LR recall = {info['recall']:.1f}%"
          if info and info['recall'] is not None else "")
    return f"Minority = {pct}%  \u00b7  trivial-baseline acc = {triv:.3f}%, recall = 0%{lr}"

init_pct = minority_pcts[0]
init_min = TOTAL_N * init_pct / 100
init_maj = TOTAL_N - init_min
fig = go.Figure()
fig.add_trace(go.Bar(x=[init_maj], y=['Transactions'], orientation='h', name='Majority',
                     marker=dict(color='#888'), hovertemplate='Majority<extra></extra>'))
fig.add_trace(go.Bar(x=[init_min], y=['Transactions'], orientation='h', name='Minority',
                     marker=dict(color=MASAI_RED), hovertemplate='Minority<extra></extra>'))

frames = []
for pct in minority_pcts:
    n_min = TOTAL_N * pct / 100
    n_maj = TOTAL_N - n_min
    frames.append(go.Frame(
        name=f'{pct}',
        data=[
            go.Bar(x=[n_maj], y=['Transactions'], orientation='h', marker=dict(color='#888')),
            go.Bar(x=[n_min], y=['Transactions'], orientation='h', marker=dict(color=MASAI_RED)),
        ],
        layout=go.Layout(
            title=_ratio_title(pct),
            annotations=[dict(
                text=_ratio_annot_text(pct),
                xref='paper', yref='paper', x=0.99, y=-0.45, xanchor='right', yanchor='bottom',
                showarrow=False, align='left', font=dict(family='monospace', size=11),
                bgcolor='#fce4ec', bordercolor=MASAI_RED, borderwidth=1, borderpad=8,
            )],
        ),
    ))

fig.frames = frames
fig.update_layout(
    barmode='stack',
    title=_ratio_title(init_pct),
    template='plotly_white', height=440,
    margin=dict(l=80, r=40, t=60, b=300),
    transition=dict(duration=600, easing='cubic-in-out'),
    annotations=[dict(
        text=_ratio_annot_text(init_pct),
        xref='paper', yref='paper', x=0.99, y=-0.45, xanchor='right', yanchor='bottom',
        showarrow=False, align='left', font=dict(family='monospace', size=11),
        bgcolor='#fce4ec', bordercolor=MASAI_RED, borderwidth=1, borderpad=8,
    )],
    sliders=[dict(
        active=0, x=0.05, y=-0.05, len=0.9, pad=dict(t=20, b=10),
        currentvalue=dict(prefix='Minority %: ', font=dict(size=14, color=MASAI_RED)),
        steps=[dict(method='animate', label=f'{pct}%',
                    args=[[f'{pct}'], dict(mode='immediate',
                                            frame=dict(duration=600, redraw=True),
                                            transition=dict(duration=600, easing='cubic-in-out'))])
               for pct in minority_pcts],
    )],
)
fig.show()
````

### 1.3 — The naive baseline (accuracy that lies)

We use the real `creditcard_50k.csv` data restricted to two real fraud features (auto-selected at ranks 10 and 15 by F-score). Two real features (instead of all 30) lets us plot the decision boundary in the plane, and picking moderately-predictive columns means the naive baseline has visibly low recall \u2014 so the 'accuracy lies' lesson lands cleanly. Same trap, real fraud data.


````python
# Reuse the real-data 2-feature split from the precompute zone
print(f"Train: {X_tr.shape}, positives: {int(y_tr.sum())}")
print(f"Test : {X_te.shape}, positives: {int(y_te.sum())}")
````

### 1.4 — Why precision and recall stay honest

Look at the confusion matrix above. The **TN cell (top-left)** dominates everything else combined. **Accuracy** divides by the total — so that one cell determines accuracy. **Precision** and **recall** never look at TN.

- Precision = TP / (TP + FP) — uses only the predicted-positive *column*.
- Recall = TP / (TP + FN) — uses only the actual-positive *row*.

When TN explodes (extreme imbalance), accuracy explodes upward. Precision and recall stay where they should.

### 2.1 — The same model on two metrics

### 2.2 — Why the two metrics disagree (Saito-Rehmsmeier 2015)

ROC plots **True Positive Rate** vs **False Positive Rate**. PR plots **Precision** vs **Recall**. Both sweep across the decision threshold from 0 to 1. The deep difference is in the **baselines**:

- **ROC-AUC's random-classifier baseline is always 0.5** — it does not depend on class frequency.
- **PR-AUC's random-classifier baseline equals the positive class frequency `π`** — 0.01 on a 1% imbalanced dataset.

On imbalanced data, ROC-AUC = 0.6 looks like "a little better than random" but PR-AUC = 0.6 on a baseline of 0.01 is <b>enormously</b> better than random. The two metrics live on different scales.


````python
box("math", "Saito-Rehmsmeier (2015), in one sentence",
    "PR-AUC's random baseline is the positive class frequency π. ROC-AUC's random baseline is always 0.5. "
    "Therefore PR-AUC reveals the difficulty of the problem; ROC-AUC hides it.")
````

### 2.4 — When ROC, when PR

| Imbalance | Headline metric | Why |
|---|---|---|
| Balanced (40–60% minority) | ROC-AUC, PR-AUC, F1 — all agree | Either works |
| Mild (10–40%) | PR-AUC preferred | More sensitive |
| Severe (&lt; 5%) | **PR-AUC mandatory** | ROC inflates |
| Extreme (&lt; 1%) | **PR-AUC + threshold tuning** | ROC misleading; default 0.5 wrong |

Default rule going forward: **on any imbalanced classification problem, report PR-AUC first.**

### 3.1 — The math behind `'balanced'`

Sklearn's `class_weight='balanced'` uses the formula:

$$w_c = \frac{n_{\text{samples}}}{n_{\text{classes}} \cdot n_c}$$

Let's compute it manually for our 99/1 dataset, then call sklearn and verify.


````python
n_samples = len(y_tr)
n_classes = 2
n_neg = int((y_tr == 0).sum())
n_pos = int((y_tr == 1).sum())

w_neg = n_samples / (n_classes * n_neg)
w_pos = n_samples / (n_classes * n_pos)

print(f"Manual derivation:")
print(f"  w_neg = {n_samples} / (2 × {n_neg}) = {w_neg:.3f}")
print(f"  w_pos = {n_samples} / (2 × {n_pos}) = {w_pos:.3f}")
print(f"  ratio = w_pos / w_neg = {w_pos/w_neg:.1f}×")
````

````python
from sklearn.utils.class_weight import compute_class_weight
weights = compute_class_weight(class_weight="balanced", classes=np.array([0,1]), y=y_tr)
print(f"sklearn says: w_neg={weights[0]:.3f}, w_pos={weights[1]:.3f}  (matches the manual computation above)")
box("math", "What this means at training time",
    f"Each minority error costs <b>{w_pos/w_neg:.0f}× more</b> than each majority error during gradient descent. "
    "The optimizer is forced to care about the rare cases.")
````

### 3.2 — Apply `class_weight='balanced'` and watch the metrics shift

````python
clf_w = LogisticRegression(max_iter=1000, class_weight="balanced", random_state=RNG).fit(X_tr, y_tr)
y_pred_w = clf_w.predict(X_te)
y_proba_w = clf_w.predict_proba(X_te)[:, 1]

metrics_w = {
    "Accuracy":  accuracy_score(y_te, y_pred_w),
    "Precision": precision_score(y_te, y_pred_w, zero_division=0),
    "Recall":    recall_score(y_te, y_pred_w),
    "F1":        f1_score(y_te, y_pred_w, zero_division=0),
    "PR-AUC":    average_precision_score(y_te, y_proba_w),
}
metrics_naive = {
    "Accuracy":  acc, "Precision": prec, "Recall": rec, "F1": f1, "PR-AUC": ap,
}

comp = pd.DataFrame({"Naive": metrics_naive, "class_weight='balanced'": metrics_w}).round(3)
comp
````

### 3.4 — `sample_weight` for per-row weighting

Class weights apply uniformly within a class. Sometimes individual rows matter more (e.g., bigger transaction amounts in fraud detection). Use `sample_weight` for that.


````python
# Synthetic 'transaction amount' to demo sample_weight
rng = np.random.default_rng(RNG)
amount = rng.lognormal(mean=8, sigma=1.2, size=len(X_tr))   # ₹ amounts
high_amount_mask = amount > np.percentile(amount, 90)       # top 10% biggest transactions

sample_w = np.where(y_tr == 1, 10.0, 1.0)         # base 10× minority weighting
sample_w[high_amount_mask] *= 5.0                  # 5× more weight to high-value rows

clf_sw = LogisticRegression(max_iter=1000, random_state=RNG).fit(X_tr, y_tr, sample_weight=sample_w)
y_pred_sw = clf_sw.predict(X_te)
print(f"With sample_weight (10× minority + 5× for top-10% amounts):")
print(f"  Recall   : {recall_score(y_te, y_pred_sw):.3f}")
print(f"  Precision: {precision_score(y_te, y_pred_sw, zero_division=0):.3f}")
print(f"  F1       : {f1_score(y_te, y_pred_sw, zero_division=0):.3f}")
box("definition", "class_weight × sample_weight is multiplicative",
    "Final weight per training sample = class_weight[y_i] × sample_weight[i]. "
    "Use <code>class_weight</code> for class-level adjustment; <code>sample_weight</code> for row-level adjustment on top.")
````

### 4.1 — The leakage anti-pattern (what NOT to do)

Below is what you might be tempted to write the first time you reach for SMOTE. **It is wrong.** Read the comments before running.


````python
# ⚠️ WRONG — SMOTE applied to the FULL dataset BEFORE the train/test split.
# Synthetic minority points derived from points the model is about to be trained on
# leak into the test set. Test scores will be inflated; production will collapse.
#
# We DEMONSTRATE the structure here but do NOT use this pattern downstream.
#
#   X_smote, y_smote = SMOTE().fit_resample(X_syn, y_syn)
#   X_tr_w, X_te_w, y_tr_w, y_te_w = train_test_split(X_smote, y_smote, stratify=y_smote)
#   clf.fit(X_tr_w, y_tr_w); clf.score(X_te_w, y_te_w)   # ← dishonest

box("danger", "SMOTE before split = data leakage",
    "Any resampling MUST happen <b>inside</b> the cross-validation / training fold, never on the full dataset before splitting. "
    "Use <code>imblearn.pipeline.Pipeline</code> so the resampling step is applied at <code>.fit()</code> only, not at <code>.predict()</code>.")
````

### 4.2 — The corrected pipeline pattern (the only correct way)

````python
pipe_smote = ImbPipeline([
    ("scaler", StandardScaler()),
    ("smote",  SMOTE(random_state=RNG, k_neighbors=5)),
    ("clf",    LogisticRegression(max_iter=1000, random_state=RNG)),
])
pipe_smote.fit(X_tr, y_tr)
y_pred_smote = pipe_smote.predict(X_te)
y_proba_smote = pipe_smote.predict_proba(X_te)[:, 1]

print("SMOTE inside ImbPipeline (correct pattern):")
print(f"  Recall   : {recall_score(y_te, y_pred_smote):.3f}")
print(f"  Precision: {precision_score(y_te, y_pred_smote, zero_division=0):.3f}")
print(f"  F1       : {f1_score(y_te, y_pred_smote, zero_division=0):.3f}")
print(f"  PR-AUC   : {average_precision_score(y_te, y_proba_smote):.3f}")
box("definition", "ImbPipeline = drop-in replacement for sklearn Pipeline",
    "<code>imblearn.pipeline.Pipeline</code> is identical to sklearn's <code>Pipeline</code> except that resampling steps "
    "(SMOTE, ADASYN, RandomUnderSampler, etc.) are <b>skipped at predict time</b>. This is the only correct pattern.")
````

### 4.3 — The SMOTE algorithm in three steps

For each minority sample $x_i$:

1. Find the $k$ nearest **minority** neighbours (Euclidean; default $k=5$).
2. Pick one neighbour $x_j$ at random.
3. Generate the synthetic point: $x_{\text{new}} = x_i + \lambda(x_j - x_i)$, where $\lambda \sim U(0, 1)$.

The synthetic point is a **linear interpolation between two real minority points**.


````python
box("definition", "SMOTE reuses the KNN distance from Session 10.1",
    "Step 1 of SMOTE is a k-nearest-neighbour search using Euclidean distance — the same computation we did in Session 10.1. "
    "Step 3 is a convex combination, the same operation that happens during gradient descent. <b>SMOTE reuses concepts; it doesn't introduce new ones.</b>")
````

### 4.4 — ADASYN difference: focus on the boundary

ADASYN does the same interpolation but **weights the number of synthetic samples per minority point by local difficulty**:

$$r_i = \frac{\#\,\text{majority neighbours in the } k\text{-neighbourhood of } x_i}{k}$$

Higher $r_i$ → more synthetics for that point. Effect: ADASYN focuses synthesis on the **decision boundary**; SMOTE spreads synthesis uniformly across minority space.

## Section 5 — Fix #3: Threshold Tuning + Multi-class Strategies

### 5.1 — F1-optimal threshold (the cheapest fix in ML)

`predict()` uses the 0.5 threshold by default. On imbalanced data, the F1-optimal threshold is usually 0.20–0.40. Tuning the threshold is **free**: no retraining, just a slider on `predict_proba`.


````python
p_arr, r_arr, t_arr = precision_recall_curve(y_te, y_proba_smote)
# Note: precision_recall_curve returns one extra (precision, recall) pair beyond thresholds.
f1_arr = 2 * p_arr[:-1] * r_arr[:-1] / (p_arr[:-1] + r_arr[:-1] + 1e-12)
best_idx = int(np.argmax(f1_arr))
best_thr = float(t_arr[best_idx])
best_f1 = float(f1_arr[best_idx])
print(f"F1-optimal threshold = {best_thr:.3f}   (F1 = {best_f1:.3f})")
print(f"Default threshold     = 0.500   (F1 = {f1_score(y_te, (y_proba_smote >= 0.5).astype(int), zero_division=0):.3f})")
box("tip", "What changed",
    f"By moving the threshold from 0.500 to <b>{best_thr:.3f}</b>, F1 improved from "
    f"<b>{f1_score(y_te, (y_proba_smote >= 0.5).astype(int), zero_division=0):.3f}</b> to <b>{best_f1:.3f}</b> — free performance.")
````

### 5.2 — Threshold for a stated business constraint (≥95% recall)

````python
recall_target = 0.95
mask_95 = r_arr[:-1] >= recall_target
if mask_95.any():
    idx_95 = int(np.where(mask_95)[0][-1])  # rightmost (largest) threshold meeting the floor
    thr_95 = float(t_arr[idx_95])
    prec_95 = float(p_arr[idx_95])
    print(f"Threshold for ≥95% recall: {thr_95:.3f}")
    print(f"  At that threshold: precision = {prec_95:.3f}, recall = {r_arr[idx_95]:.3f}")
    box("tip", "Translate to business language",
        f"At threshold <b>{thr_95:.3f}</b> we catch <b>{r_arr[idx_95]:.0%} of fraud</b>, "
        f"but only <b>{prec_95:.0%}</b> of flagged transactions are real fraud — meaning ~{(1-prec_95)/prec_95:.1f} false alarms per real fraud. "
        "That's the conversation to have with the head of fraud investigations.")
else:
    print("No threshold achieves ≥95% recall on this model. Train a stronger model first.")
````

### 5.4 — Multi-class: OvR vs OvO

When you have more than two classes, two standard wrappers exist:

- **One-vs-Rest (OvR)** trains *n* binary classifiers — each one separates "is this class X" from "is this not class X". Linear in the number of classes.
- **One-vs-One (OvO)** trains $n \times (n-1)/2$ binary classifiers — one for each pair of classes. Quadratic.

For 10 classes, OvR trains 10 models; OvO trains 45.


````python
wine = load_wine()
print(f"Wine dataset: {wine.data.shape[0]} samples, {len(wine.target_names)} classes ({list(wine.target_names)})")

ovr = OneVsRestClassifier(SVC(random_state=RNG)).fit(wine.data, wine.target)
ovo = OneVsOneClassifier(SVC(random_state=RNG)).fit(wine.data, wine.target)
print(f"OvR trained {len(ovr.estimators_)} classifiers (= n classes)")
print(f"OvO trained {len(ovo.estimators_)} classifiers (= n(n-1)/2)")

# Hypothetical scaling for n=10 classes
n_demo = 10
print(f"\nIf instead we had {n_demo} classes:")
print(f"  OvR: {n_demo} classifiers")
print(f"  OvO: {n_demo*(n_demo-1)//2} classifiers — the quadratic blowup")
````

````python
box("definition", "Sklearn defaults",
    "<code>LogisticRegression</code> defaults to multinomial / OvR-style softmax. <code>SVC</code> defaults to OvO. "
    "For linear models on big data, OvR is the right choice. For kernel SVMs on smaller data, OvO often wins.")
````

## Section 6 — Mini-Project: Final Pipeline on Real Fraud Data

Bring it all together. Target: F1 ≥ 0.85 on the held-out fraud test set.

### 6.1 — Load and verify the dataset

````python
df = pd.read_csv("creditcard_50k.csv")
print(f"Loaded: {df.shape[0]:,} rows, {df.shape[1]} columns")
print(f"Class distribution:\n{df['Class'].value_counts()}")
fraud_rate = df["Class"].mean()
print(f"Fraud rate: {fraud_rate*100:.4f}%")
assert 0.001 < fraud_rate < 0.005, f"Unexpected fraud rate {fraud_rate:.4%} — expected ~0.172%"
print("Dataset verified.")
````

### 6.2 — Stratified train/test split

````python
X = df.drop(columns=["Class"])
y = df["Class"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, stratify=y, test_size=0.2, random_state=RNG,
)
print(f"Train: {len(X_tr):,} rows ({int(y_tr.sum())} frauds = {y_tr.mean()*100:.3f}%)")
print(f"Test : {len(X_te):,} rows ({int(y_te.sum())} frauds = {y_te.mean()*100:.3f}%)")
n_minority_train = int(y_tr.sum())
smote_k = min(5, max(1, n_minority_train - 1))
print(f"\nSMOTE k_neighbors will be set to {smote_k} (must be < n_minority_train = {n_minority_train})")
````

### 6.3 — The final pipeline

````python
final_pipe = ImbPipeline([
    ("scaler", StandardScaler()),
    ("smote",  SMOTE(random_state=RNG, k_neighbors=smote_k)),
    ("clf",    LogisticRegression(max_iter=2000, C=0.1, random_state=RNG)),
])
final_pipe.fit(X_tr, y_tr)
final_proba = final_pipe.predict_proba(X_te)[:, 1]
print("Pipeline fitted.")
````

### 6.4 — Threshold-tune and evaluate

````python
p_f, r_f, t_f = precision_recall_curve(y_te, final_proba)
f1_f = 2 * p_f[:-1] * r_f[:-1] / (p_f[:-1] + r_f[:-1] + 1e-12)
best_idx_f = int(np.argmax(f1_f))
final_thr = float(t_f[best_idx_f])
y_pred_final = (final_proba >= final_thr).astype(int)

final_f1   = f1_score(y_te, y_pred_final, zero_division=0)
final_prec = precision_score(y_te, y_pred_final, zero_division=0)
final_rec  = recall_score(y_te, y_pred_final)
final_ap   = average_precision_score(y_te, final_proba)

print("=" * 50)
print("FINAL PIPELINE RESULTS (held-out fraud test set)")
print("=" * 50)
print(f"  Threshold (F1-optimal) : {final_thr:.4f}")
print(f"  F1 on fraud class      : {final_f1:.4f}")
print(f"  Precision              : {final_prec:.4f}")
print(f"  Recall                 : {final_rec:.4f}")
print(f"  PR-AUC (Average Prec)  : {final_ap:.4f}")
print()
print(classification_report(y_te, y_pred_final, target_names=["legit", "fraud"]))
````

### 6.7 — 10 Key Takeaways

1. **`y.value_counts(normalize=True)` is line one** of every classification project.
2. **Accuracy on imbalanced data is meaningless.** A 99.83%-accurate fraud model can detect zero frauds.
3. **Use PR-AUC as the headline metric for imbalanced binary classification.** ROC baseline always 0.5; PR baseline = class frequency.
4. **The default 0.5 threshold is almost never optimal** on imbalanced data. Tune to F1 or to a stated business constraint.
5. **`class_weight='balanced'`** uses the formula `n_samples / (n_classes × n_class)`. Custom `{0:1, 1:k}` is preferable when you have a stated cost ratio.
6. **Resampling MUST happen inside `imblearn.pipeline.Pipeline`.** SMOTE-before-split is data leakage.
7. **SMOTE = KNN + linear interpolation.** Reuses Session 10.1's distance idea.
8. **ADASYN focuses on the boundary**; SMOTE spreads uniformly across minority space.
9. **Random under-sampling throws information away.** Useful as a baseline; rarely a final answer.
10. **OvR is linear in classes; OvO is quadratic.** sklearn LogReg defaults to multinomial; SVC defaults to OvO.

### 6.9 — Common Pitfalls Quick Reference

| Symptom | Root cause | Fix |
|---|---|---|
| Model says 99% accuracy but business hates it | Accuracy on imbalanced data is meaningless | Use precision / recall / F1 / PR-AUC |
| PR-AUC = 0.99 in CV, 0.4 in production | SMOTE applied before split = data leakage | SMOTE inside `ImbPipeline` |
| `class_weight='balanced'` made precision crater | Recall went up, precision went down by design | Tune threshold or use a smaller cost ratio |
| ROC-AUC = 0.95 but model fails on minority | ROC is forgiving on imbalance | Use PR-AUC (`average_precision_score`) |
| `ValueError: n_neighbors <= n_samples` from SMOTE | Train fold has fewer minorities than `k_neighbors` | Lower `k_neighbors` (try 3 or 1) |
| `precision_recall_curve` shape mismatch when computing F1 | Returns one extra `(prec, rec)` for threshold = 0 | Slice `prec[:-1]`, `rec[:-1]` |
