```python
import numpy as np
```

```python
## List to numpy array
## Numpy Array: Homogeneous Continuous Array of elements.

my_list = [1, 2, 3, 4, 5]
np_from_list = np.array(my_list)
print(my_list)
print(np_from_list)

print(type(my_list))
print(type(np_from_list))
print(np_from_list.shape)
print(np_from_list.ndim)
```

```python
## 2D Array
np_arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(np_arr_2d)
print(np_arr_2d.shape)
print(np_arr_2d.ndim)
```

```python
%%time
doubled_list = [2 * i for i in sample_list]
```

```python
## Creating Numpy Arrays
arr1 = np.zeros(4) ## Vector of 4 zeroes
arr2 = np.ones((4, 5)) ## Matrix of ones with shape [4*5]
tensor = np.ones((4, 5, 3)) ## Tensor(3D) of ones
print(arr1)
print(arr1.shape)
print(arr1.ndim)
print(arr1.dtype)

print(arr2)
print(arr2.shape)
print(arr2.ndim)
print(arr2.dtype)

print(tensor)
print(tensor.shape)
print(tensor.ndim)
print(tensor.dtype)
```

```python
arr3 = np.arange(0, 100, 2) ## Similar to range
print(arr3)
print(arr3.shape)
print(arr3.dtype)
print(arr3.ndim)
print(arr3.size)
```

```python
arr4 = np.linspace(0, 100, 25) ## Linear equispaced floats
print(arr4)
print(arr4.shape)
print(arr4.dtype)
print(arr4.ndim)
print(arr4.size)
```

```python
arr5 = np.full((3, 4), 20) ## Fill with specific values, creates an matrix of shape (3, 4) and fills it with 20
print(arr5)
print(arr5.shape)
print(arr5.ndim)
print(arr5.dtype)
print(arr5.size)
```

```python
## Mixed Types -> NumPy automatically upcasts everything to the most general/safe type that can hold all values.
print(np.array([1, 2.0, 3.5]).dtype)
print(np.array([1, 2, True]).dtype)
print(np.array([1, 2, True, 4.32]).dtype)
print(np.array([1, 2, True, "Harry"]).dtype)
print(np.array([1, 2, True, 8.0, "Jack"]).dtype)
```
