```python
import numpy as np
```

```python
## 1. Slicing -> To Access Contiguous Elements
arr = np.array([2, 4, 3, 1, 7])
print(arr[1:3])
print(arr[:4])
print(arr[2:])
print(arr[:-1])
print(arr[-4:-2])
```

```python
## Slicing in 2D Array
mat = np.array([
    [2, 4, 5],
    [6, 9, 1],
    [7, 8, 10],
    [1, 7, 3]
])
## mat[row_slice, col_slice]
print(mat[:2, 0:2])
print(mat[:, :2])
print(mat[:-1, :-1])
print(mat[0:2])
```

```python
## Fancy Indexing: Passing a list of indexes to select. The order of elements in output is same as order of input index list
print(arr[[0, 3, 2]])
print(arr[[3, 3, 2, 1, 4, 0]])
```

```python
## Matrix Fancy Indexing: mat[row_idx, col_idx]
print(mat[[0, 2], [2, 1]])
print(mat[[2, 2, 1, 0], [1, 0, 0, 2]])
print(mat[:-1, [2, 0, 1]]) ## Slicing + Fancy Indexing
```
